```python
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.datasets import load_iris
from sklearn.cluster import KMeans
from sklearn.cluster import AgglomerativeClustering

iris = load_iris()

df = pd.DataFrame(
    iris.data,
    columns=iris.feature_names
)

df.head()
```

```python
kmeans = KMeans(
    n_clusters=3,
    random_state=42
)

df["Cluster"] = kmeans.fit_predict(df)
```

```python
plt.figure(figsize=(8,6))

plt.scatter(
    df.iloc[:,0],
    df.iloc[:,1],
    c=df["Cluster"],
    cmap="viridis"
)

plt.xlabel("Sepal Length")
plt.ylabel("Sepal Width")
plt.title("K-Means Clustering on Iris Dataset")

plt.show()
```

K-Means is an unsupervised machine learning algorithm used to group similar data points into K clusters. It works by assigning each data point to the nearest centroid (center of the cluster) and updating the centroids until they no longer change.

Choose the number of clusters (K).
Randomly place K centroids.
Assign each data point to the nearest centroid.
Calculate the new centroid by taking the mean of all points in the cluster.
Repeat the process until the centroids stop changing.

K-Means is suitable for the Iris dataset because it contains numerical features such as sepal length, sepal width, petal length, and petal width. These features naturally form similar groups, making it easy for K-Means to cluster the flowers based on their measurements. The Iris dataset is also small, so K-Means works efficiently.

```python
from scipy.cluster.hierarchy import linkage, dendrogram
from sklearn.cluster import AgglomerativeClustering
linked = linkage(df.iloc[:,0:4], method="ward")

plt.figure(figsize=(10,5))
dendrogram(linked)

plt.title("Dendrogram")
plt.xlabel("Data Points")
plt.ylabel("Euclidean Distance")

plt.show()
```

```python
hc = AgglomerativeClustering(n_clusters=3)

df["HC_Cluster"] = hc.fit_predict(df.iloc[:,0:4])
```

```python
plt.figure(figsize=(8,6))

plt.scatter(
    df.iloc[:,0],
    df.iloc[:,1],
    c=df["HC_Cluster"],
    cmap="viridis"
)

plt.xlabel("Sepal Length")
plt.ylabel("Sepal Width")
plt.title("Hierarchical Clustering on Iris Dataset")

plt.show()
```

Hierarchical Clustering is an unsupervised machine learning algorithm that groups similar data points into clusters based on their distance. It creates a hierarchy of clusters by repeatedly merging the closest clusters until all data points form a single cluster. The clustering process is visualized using a dendrogram, which helps determine the optimal number of clusters.

Hierarchical Clustering is suitable for the Iris dataset because it groups flowers with similar measurements without requiring target labels. The dendrogram helps visualize the relationships between data points and makes it easier to identify the natural clusters present in the dataset. Since the Iris dataset contains three distinct flower species with similar characteristics, Hierarchical Clustering can effectively separate them into meaningful groups.
